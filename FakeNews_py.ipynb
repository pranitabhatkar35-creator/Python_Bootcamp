```python
#import the random module
import random

subjects = [
    "Shahrukh Khan",
    "A Mumbai paplet",
    "A group of monkeys",
    "Auto driver in Delhi",
    "A dog in Kolkata",
    "PM Modiji",
]

actions = [
    "launches",
    "cancels",
    "lands",
    "eats",
    "delays",
    "walks with",
]

places_or_things = [
    "at the red carpet",
    "in Delhi metro",
    "a plate of Maggi",
    "in Taj Hotel",
    "at China Gate",
    "during IPL match",
]

# Start the headline generation loop
while True:
    subject = random.choice(subjects)
    action = random.choice(actions)
    place = random.choice(places_or_things)

    headline = f"BREAKING NEWS: {subject} {action} {place}!"
    print("\n" + headline)

    user_input = input("\nDo you want another headline? (yes/no): ").strip().lower()

    if user_input == "no":
        print("\nThanks for using the fake news headline generator")
        break


```
